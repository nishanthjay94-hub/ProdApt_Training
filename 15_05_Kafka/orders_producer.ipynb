```python
# Producer Example for Kafka (new_orders topic)

from kafka import KafkaProducer

# Connect to Kafka broker (host port 9093 mapped to container 9092)
producer = KafkaProducer(
    bootstrap_servers='localhost:9093',
    value_serializer=lambda v: v.encode('utf-8')  # convert string → bytes
)

# Send a few test messages
messages = [
    "Order 101 created",
    "Order 102 shipped",
    "Order 103 delivered"
]

for msg in messages:
    producer.send('new_orders', msg)
    print(f"Sent: {msg}")

producer.flush()
print("All messages sent successfully!")
```
