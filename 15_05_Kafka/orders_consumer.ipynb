```python
# Consumer Example for Kafka (new_orders topic)

from kafka import KafkaConsumer

# Connect to Kafka broker (host port 9093 mapped to container 9092)
consumer = KafkaConsumer(
    'new_orders',
    bootstrap_servers='localhost:9093',
    auto_offset_reset='earliest',   # start from beginning if no offset
    enable_auto_commit=True,
    group_id='jupyter-group',
    value_deserializer=lambda v: v.decode('utf-8')  # convert bytes → string
)

print("Listening for messages on 'new_orders'...")

for message in consumer:
    print(f"Received: {message.value}")
    break  # stop after first message for demo
```
