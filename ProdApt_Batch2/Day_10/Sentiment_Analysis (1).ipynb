```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk

# Download VADER lexicon
nltk.download('vader_lexicon')

# Initialize the SentimentIntensityAnalyzer
sid = SentimentIntensityAnalyzer()

# Sample text for sentiment analysis
text = "I absolutely love this product! It’s amazing and works perfectly."

# Perform sentiment analysis
sentiment_scores = sid.polarity_scores(text)

# Output sentiment scores
print(sentiment_scores)

# Interpret the result
if sentiment_scores['compound'] >= 0.05:
    print("Positive sentiment")
elif sentiment_scores['compound'] <= -0.05:
    print("Negative sentiment")
else:
    print("Neutral sentiment")
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import nltk

# Download VADER lexicon
nltk.download('vader_lexicon')

# Initialize SentimentIntensityAnalyzer
sid = SentimentIntensityAnalyzer()

# List of 20 sample customer feedbacks
feedbacks = [
    "I love the product! It works perfectly.",
    "Terrible experience, would not recommend.",
    "Amazing quality and great support team.",
    "The product is okay, nothing special.",
    "Horrible! It broke within a day.",
    "Very satisfied with my purchase. Excellent value for money!",
    "Not as described, very disappointed.",
    "Fast delivery and great customer service.",
    "The item doesn't work at all.",
    "Perfect product, exceeded my expectations.",
    "The colors are dull and lifeless.",
    "So happy I bought this. Best decision ever!",
    "The item was delivered late and damaged.",
    "Great for the price, I'm happy with it.",
    "Cheap material, not worth the money.",
    "Loved the sleek design and functionality!",
    "Would buy again, highly recommended.",
    "The customer service was unhelpful and rude.",
    "It's a decent product for its price.",
    "Fantastic! Will tell my friends about it."
]

# Analyze each feedback and categorize it
print("{:<5} {:<60} {:<10}".format("No.", "Feedback", "Sentiment"))
for i, feedback in enumerate(feedbacks, start=1):
    sentiment_scores = sid.polarity_scores(feedback)
    compound_score = sentiment_scores['compound']
    
    # Categorize sentiment
    if compound_score >= 0.05:
        sentiment = "Positive"
    elif compound_score <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"
    
    print("{:<5} {:<60} {:<10}".format(i, feedback, sentiment))
```
